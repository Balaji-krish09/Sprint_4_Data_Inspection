# Notebook 2: Data Inspection


```python
import pandas as pd
# Load the dataset
df = pd.read_csv("Customer_Data.csv")
df.head()
```

### 1. Number of Rows & 2. Number of Columns
Retrieving individual axis dimensions of the dataset.

```python
print("Number of Rows:", len(df))
print("Number of Columns:", len(df.columns))
```

### 3. Data Types & 12. dtypes
Checking memory representations for each feature column.

```python
print(df.dtypes)
```

### 4. Column Names
Listing all structural features in the dataset.

```python
print(list(df.columns))
```

### 5. Index
Inspecting row indexing scheme.

```python
print(df.index)
```

### 6. head(), 7. tail(), and 8. sample()
Previewing top records, bottom records, and random entries.

```python
# Top 5 rows
df.head()

# Bottom 5 rows
df.tail()

# Random 5 rows sample
df.sample(5, random_state=42)
```

### 9. shape & 10. info()
Examining dimensional matrix shape and complete structural metadata.

```python
print("Shape:", df.shape)
print('-'*40)
df.info()
```

### 11. describe()
Generating summary statistics for both numerical and categorical attributes.

```python
# Numerical Summary
df.describe()

# Categorical Summary
df.describe(include=['object'])
```

### 13. nunique() & 14. value_counts()
Evaluating unique value counts and frequency distributions.

```python
print('--- Unique Value Counts per Column ---')
print(df.nunique())

print('\n--- Target Distribution (Churn) ---')
print(df['Churn'].value_counts())
```

| Classification Category | Column(s) Identified | Reason / Description |

| **Target Column** | `Churn` | Primary target outcome classification (`Yes` / `No`). |
