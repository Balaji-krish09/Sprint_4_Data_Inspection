### 3. Target Variable
The **Target** is the specific column we want our predictive model to forecast. Here, `Churn` is our target.

```python
# Check target distribution
print(df["Churn"].value_counts())
```

### 4. Independent & Dependent Variables
- **Independent Variables ($X$):** The predictor columns used as inputs.
- **Dependent Variable ($y$):** The outcome being predicted (`Churn`).

```python
# Separate predictors (X) and target (y)
X = df.drop(columns=["Churn"])
y = df["Churn"]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

### 5. Numerical vs Categorical Features
- **Numerical:** Quantitative counts or measurements (`Age`, `MonthlyCharges`, `TenureYears`).
- **Categorical:** Qualitative categories or text labels (`Gender`, `ContractType`, `PaymentMethod`).

```python
# Identify numerical vs categorical
num_cols = df.select_dtypes(include=["number"]).columns
cat_cols = df.select_dtypes(include=["object"]).columns

print("Numerical Columns:", list(num_cols))
print("Categorical Columns:", list(cat_cols))
```

### 6. Metadata & Dimensions
Metadata provides structural information including data types, non-null counts, and shape.
