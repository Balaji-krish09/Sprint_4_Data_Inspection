# Notebook 12: Business Insights & Strategic Action Plan

#### Overview
Convert Exploratory Data Analysis (EDA) findings into actionable business insights. This notebook moves from **Visualization $\rightarrow$ Observation $\rightarrow$ Strategic Insight** to identify patterns, trends, relationships, anomalies, business problems, key customer segments, and high-value opportunities.

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

sns.set_theme(style="whitegrid")

# Load dataset
df = pd.read_csv("Customer_Data.csv")
df.head()
```

### 1. Moving from Visualization → Observation → Insight

#### Bad Example (Pure Visualization Description):
"The bar chart shows churn counts across contract types."

#### Good Example (Visualization → Observation → Business Insight):
"Month-to-month contract holders show a churn rate of nearly 42%, whereas 1-year and 2-year contract holders churn at under 10%. While month-to-month users represent a large subscriber base, they create a severe revenue leakage point. Structuring promotional discounts for annual contract conversions will directly improve long-term retention."

```python
# Analyzing Contract Type vs Churn Percentage
contract_churn = (
    pd.crosstab(df["ContractType"], df["Churn"], normalize="index") * 100
).round(2)
print("=== Churn Percentage by Contract Type (%) ===")
print(contract_churn)

# Plotting
plt.figure(figsize=(8, 4))
sns.countplot(
    data=df, x="ContractType", hue="Churn", palette="Set1"
)
plt.title("Contract Type vs Churn Count")
plt.show()
```

### 2. Business Insights & Opportunity Matrix

| Focus Area | Observed Pattern / Trend | Potential Business Problem | Strategic Opportunity |
| :--- | :--- | :--- | :--- |
| **Contract Type** | Month-to-month contracts account for over 75% of total churned customers. | High acquisition cost wasted on short-lived subscribers. | Offer incentive discounts for switching to 1-year terms. |
| **Payment Method** | Electronic check users exhibit double the churn rate of auto-pay users. | Payment friction or manual billing drops. | Incentivize automated credit card/bank transfer enrollment with a one-time bill credit. |
| **Internet Service** | Fiber optic customers churn at ~35.8% despite paying higher monthly rates. | Value-to-price gap or service dissatisfaction among high-value tiers. | Priority technical support and quality-of-service audits for Fiber optic subscribers. |
