# Notebook 2: Data Inspection

#### Overview
Perform deep inspection on `Customer_Data.csv` including structural dimensions, data types, statistical previews, and column classification.

```python
import pandas as pd

# Load the dataset
df = pd.read_csv("Customer_Data.csv")
df.head()
```

### 1. Number of Rows & 2. Number of Columns
Retrieving individual axis dimensions of the dataset.

```python
print("Number of Rows:", len(df))
print("Number of Columns:", len(df.columns))
```

### 3. Data Types & dtypes
Inspecting low-level memory representations and Python data types for every feature.

```python
print("--- Column Data Types ---")
print(df.dtypes)
```

### 4. Column Names & 5. Index
Checking structural feature headers and row indexing boundaries.

```python
print("Column Names:", list(df.columns))
print("\nDataset Index:", df.index)
```

### 6. head(), 7. tail(), and 8. sample()
Inspecting top records, bottom records, and random subsets to catch data anomalies.

```python
# View first 5 rows
df.head()

# View last 5 rows
df.tail()

# View 5 randomly sampled rows
df.sample(5, random_state=42)
```
